```python
5 in [5, 3, 0, 7]
4 in [5, 3, 0, 7]


def is_valid(puzzle, row, col, num):
    if num in puzzle[row]:
        return False
    col_values = []
    for r in range(9):
        col_values.append(puzzle[r][col])
    if num in col_values:
        return False

    box_row = row // 3 * 3
    box_col = col // 3 * 3

    box = []
    for r in range(box_row, box_row + 3):
        for c in range(box_col, box_col + 3):
            box.append(puzzle[r][c])
    if num in box:
        return False
    return True


print(is_valid(puzzle, 0, 2, 4))
print(is_valid(puzzle, 0, 2, 5))
```

```python
def find_empty(puzzle):
    for r in range(9):
        for c in range(9):
            if puzzle[r][c] == 0:
                return r, c
    return None
```

```python
def solve(puzzle):
    empty = find_empty(puzzle)
    if empty is None:
        return True

    r, c = empty
    for num in range(1, 10):
        if is_valid(puzzle, r, c, num):
            puzzle[r][c] = num

            if solve(puzzle):
                return True

            puzzle[r][c] = 0
    return False

solve(puzzle)
for row in puzzle:
    print(row)

```
